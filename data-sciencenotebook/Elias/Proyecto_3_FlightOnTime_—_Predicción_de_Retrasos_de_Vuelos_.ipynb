```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Ingresando dataframe a la variable datos

```python
datos = pd.read_csv('/content/flights_sample_3m.csv', engine='python');
datos.head()
```

```python
datos.info()
```

```python
datos.describe(include = "all")
```

Limpiando Datos

```python
datos.isna().sum().sort_values(ascending=False)
```

```python
datos.isna().sum().sort_values(ascending=False) * 100
```

```python
datos['DEP_DELAY'] = datos['DEP_DELAY'].fillna(0)
```

```python
datos[datos['CANCELLED'] == 1].isna().sum()
```

Separar los vuelos separados

```python
vuelos_cancelados = datos.loc[datos['CANCELLED'] == 1].copy()
vuelos_operados = datos.loc[datos['CANCELLED'] == 0].copy()
```

```python
# Limpiando fechas invalidas
vuelos_operados.loc[:, 'FL_DATE'] = pd.to_datetime(vuelos_operados['FL_DATE'], errors='coerce')
vuelos_operados[vuelos_operados['FL_DATE'].isna()]
```

```python
#retrasos
vuelos_operados[
    (vuelos_operados['DEP_DELAY'] < -60) |
    (vuelos_operados['DEP_DELAY'] > 1000)
]
```

```python
vuelos_operados[
    (vuelos_operados['DISTANCE'] <= 0) |
    (vuelos_operados['DISTANCE'] > 15000)
]
```

## Tipo de dato distancia

#### La variable distance se encuentra el millas

```python
vuelos_operados['DISTANCE'].describe()
```

## Convirtiendo a Kilometros

```python
vuelos_operados['DISTANCE_KM'] = vuelos_operados['DISTANCE'] * 1.60934
vuelos_operados['DISTANCE_KM'].describe()
```

## Graficando

### Cancelaciones por aerolinea (en caso de que sea relevante)

```python
tasa_cancelacion = (
    datos.groupby('AIRLINE')['CANCELLED'].mean().sort_values(ascending=False)
)
tasa_cancelacion.plot(kind='bar')
```

```python
plt.figure(figsize=(10,8))
sns.barplot(
    y=tasa_cancelacion.index,
    x=tasa_cancelacion.values,
    orient='h'
)
plt.title("Cancelación por aerolinea")
plt.xlabel("Vuelos cancelados")
plt.ylabel("Aerolinea")
plt.tight_layout()
plt.show()
```

### Retrasos por aerolinea

```python
top_aerolineas = (vuelos_operados['AIRLINE'].value_counts().head(10).index)

datos_top = vuelos_operados[vuelos_operados['AIRLINE'].isin(top_aerolineas)].copy()

datos_top.loc[:, 'DEP_DELAY_CLIPPED'] = datos_top['DEP_DELAY'].clip(lower=-60, upper=180)

plt.figure(figsize=(10,6))
sns.boxplot(data=datos_top, y='AIRLINE', x='DEP_DELAY_CLIPPED')
plt.title('Retrasos por aerolinea (Top 10)')
plt.tight_layout()
plt.show()
```

```python
# En graficas en vertical

plt.figure(figsize=(14,7))
sns.barplot(x=tasa_cancelacion.index, y=tasa_cancelacion.values)
plt.xticks(rotation=90)
plt.title('Cancelación por aerolinea')
plt.tight_layout()
plt.show()
```

## cancelación en relacion con retrasos

```python
tasa_cancelacion_promedio = (
    datos.groupby('AIRLINE')['CANCELLED'].mean()
)

retraso_promedio = (
    datos[datos['CANCELLED']==0].groupby('AIRLINE')['DEP_DELAY'].mean()
)
```

```python
cancelaciones_retrasos = pd.concat(
    [tasa_cancelacion_promedio, retraso_promedio],
    axis=1
).reset_index()

cancelaciones_retrasos.columns = ['AIRLINE', 'tasa_cancelacion_promedio', 'retraso_promedio']

cancelaciones_retrasos.describe()
```

```python
cancelaciones_retrasos[['tasa_cancelacion_promedio', 'retraso_promedio']].corr()
```

## Graficando cancelaciones y retrasos

```python
cancelaciones_retrasos_top = cancelaciones_retrasos[cancelaciones_retrasos['AIRLINE'].isin(top_aerolineas)]
```

```python
fig, ax1 = plt.subplots(figsize=(12,6))

x = range(len(cancelaciones_retrasos_top))

ax1.bar(
    x,
    cancelaciones_retrasos_top['tasa_cancelacion_promedio'],
    alpha=0.7,
)

ax1.set_ylabel('Cancelaciones promedio')

ax1.set_xticks(x)
ax1.set_xticklabels(
    cancelaciones_retrasos_top['AIRLINE'],
    rotation=45,
    ha='right'
  )

ax2 = ax1.twinx()
ax2.plot(
    x,
    cancelaciones_retrasos_top['retraso_promedio'],
    color='orange',
    marker='o',
)
ax2.set_ylabel('Retraso promedio (en minutos)')

plt.title("Cancelaciones vs retraso por aerolinea")
fig.tight_layout()
plt.show()
```
